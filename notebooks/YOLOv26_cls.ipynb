```python
from ultralytics import YOLO
```

```python
model = YOLO('yolo26s-cls.pt')
```

```python
model.info()
```

```python
result = model.train(
    data="../Final_v26_cls/",
    epochs=40,
    imgsz=640,
    batch=8,        
    device=0, 
    workers=2,
    amp=True,    
    cache=False, 
    patience=10,
)
```
